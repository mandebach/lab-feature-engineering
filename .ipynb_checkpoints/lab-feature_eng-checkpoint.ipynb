```python
# Step 0 - Libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import SGDClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

SEED = 1
```

```python
# Step 1 - Load Dataset and EDA
spaceship = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/spaceship_titanic.csv")
spaceship.head()
```

**Check the shape of your data**

```python
spaceship.shape
```

**Check for data types**

```python
spaceship.info()
```

**Check for missing values**

```python
spaceship.isnull().sum()
```

There are multiple strategies to handle missing data

- Removing all rows or all columns containing missing data.
- Filling all missing values with a value (mean in continouos or mode in categorical for example).
- Filling all missing values with an algorithm.

For this exercise, because we have such low amount of null values, we will drop rows containing any missing value. 

```python
spaceship_clean = spaceship.dropna()
```

```python
spaceship_clean.isnull().sum()
```

- **Cabin** is too granular - transform it in order to obtain {'A', 'B', 'C', 'D', 'E', 'F', 'G', 'T'}

```python
spaceship_clean["Cabin"] = spaceship_clean["Cabin"].str[0]
```

```python
spaceship_clean.head()
```

- Drop PassengerId and Name

```python
spaceship_clean = spaceship_clean.drop(columns=["PassengerId", "Name"])
```

```python
spaceship_clean.head()
```

- For non-numerical columns, do dummies.

```python
spaceship_clean["HomePlanet"].value_counts()
```

```python
spaceship_clean["Destination"].value_counts()
```

```python
spaceship_clean["Cabin"].value_counts()
```

```python
encoder = OneHotEncoder(sparse_output=False, drop='first')

cols_to_encode = ["HomePlanet", "Destination", "Cabin"]
spaceship_encoded = encoder.fit_transform(spaceship_clean[cols_to_encode])

encoded_cols = encoder.get_feature_names_out(cols_to_encode)

spaceship_encoded_df = pd.DataFrame(spaceship_encoded, columns=encoded_cols, index=spaceship_clean.index)

spaceship_encoded_df.head()
```

```python
# Consolidate into dataset

spaceship_features = pd.concat([
    spaceship_clean[['CryoSleep', 'Age', 'VIP', 'RoomService', 'FoodCourt', 'ShoppingMall', 'Spa', 'VRDeck', 'Transported' ]],
    spaceship_encoded_df
], axis=1)


spaceship_features.head()
```

**Perform Train Test Split**

```python
# Step 2 - Train / Test Split

# Select the features (X) and the target (y)
X = spaceship_features.drop(columns=["Transported"])
y = spaceship_features['Transported']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,y,
    test_size = 0.2,
    random_state=SEED,
    stratify=y)

# To see the shape of the Train (80%) and Test (20%) datasets
print(f" X_train shape: {X_train.shape}")
print(f" X_test shape: {X_test.shape}")
```

```python
# Step 3 - Feature Engineering: Feature Scaling

scaler = StandardScaler()
scaler.set_output(transform="pandas")

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

**Model Selection**

In this exercise we will be using **KNN** as our predictive model.

```python
# Step 4 - Model training (KNN)

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train_scaled, y_train)

print("Model trained!")
```

```python
y_pred = knn.predict(X_test_scaled)

print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}")
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```
